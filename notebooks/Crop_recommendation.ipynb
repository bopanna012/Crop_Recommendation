```python
## Split data into x and y
x = df.drop('label', axis=1)
y = df.label
```

```python
## training and testing data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

### **Train-Test Split**
The dataset is split into:
- 80% training data
- 20% testing data

Training data is used to build the model, while testing data is used to evaluate performance.
This helps prevent overfitting and improves generalization.

### **Logistic Regression Model**
Logistic Regression is a supervised classification algorithm
used as a baseline model for crop prediction.

The model learns the relationship between input features and crop labels. It predicts the probability of different crop classes based on input features.

The accuracy score indicates how well the model predicts the correct crop.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

# Logistic Regression is scale-sensitive, so it is wrapped in a Pipeline with
# StandardScaler to give it a fair comparison against the tree-based models below.
logistic_regression_model = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])
logistic_regression_model.fit(x_train, y_train)

logistic_regression_pred = logistic_regression_model.predict(x_test)

conf_matrix = confusion_matrix(y_test, logistic_regression_pred)
print("Confusion Matrix:")
print(conf_matrix)

accuracy = accuracy_score(y_test, logistic_regression_pred)
precision = precision_score(y_test, logistic_regression_pred, average="weighted", zero_division=0)
recall = recall_score(y_test, logistic_regression_pred, average="weighted", zero_division=0)
f1 = f1_score(y_test, logistic_regression_pred, average="weighted", zero_division=0)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

### **Logistic Regression Insight**
The performance of the Logistic Regression model is evaluated using:
- Accuracy
- Precision
- Recall
- F1-Score

These metrics help measure how well the model predicts the correct crop.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

random_forest_model = RandomForestClassifier(random_state=42)
random_forest_model.fit(x_train, y_train)

random_forest_pred = random_forest_model.predict(x_test)

conf_matrix = confusion_matrix(y_test, random_forest_pred)
print("Confusion Matrix:")
print(conf_matrix)

accuracy = accuracy_score(y_test, random_forest_pred)
precision = precision_score(y_test, random_forest_pred, average="weighted", zero_division=0)
recall = recall_score(y_test, random_forest_pred, average="weighted", zero_division=0)
f1 = f1_score(y_test, random_forest_pred, average="weighted", zero_division=0)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

gradient_boosting_model = GradientBoostingClassifier(random_state=42)
gradient_boosting_model.fit(x_train, y_train)

gradient_boosting_pred = gradient_boosting_model.predict(x_test)

conf_matrix = confusion_matrix(y_test, gradient_boosting_pred)
print("Confusion Matrix:")
print(conf_matrix)

accuracy = accuracy_score(y_test, gradient_boosting_pred)
precision = precision_score(y_test, gradient_boosting_pred, average="weighted", zero_division=0)
recall = recall_score(y_test, gradient_boosting_pred, average="weighted", zero_division=0)
f1 = f1_score(y_test, gradient_boosting_pred, average="weighted", zero_division=0)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

### **Support Vector Machine Model**
Support Vector Machine is a powerful classification algorithm.

It works by finding the optimal boundary that separates different crop classes. It performs well when the relationship between features is complex.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

# SVMs are distance-based, so features are scaled the same way as for Logistic Regression.
svc_model = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", SVC(random_state=42))
])
svc_model.fit(x_train, y_train)

svc_pred = svc_model.predict(x_test)

conf_matrix = confusion_matrix(y_test, svc_pred)
print("Confusion Matrix:")
print(conf_matrix)

accuracy = accuracy_score(y_test, svc_pred)
precision = precision_score(y_test, svc_pred, average="weighted", zero_division=0)
recall = recall_score(y_test, svc_pred, average="weighted", zero_division=0)
f1 = f1_score(y_test, svc_pred, average="weighted", zero_division=0)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

### **Saving the Final Model**
Random Forest is persisted with `joblib` so the Streamlit app (`app/app.py`) can
load it directly, keeping the notebook and the deployed model in sync.

```python
import os
import joblib

MODELS_DIR = os.path.join("..", "models")
os.makedirs(MODELS_DIR, exist_ok=True)

# Random Forest had the best accuracy/F1 of the four candidates, so it is the
# model saved for the Streamlit app in app/app.py.
joblib.dump(random_forest_model, os.path.join(MODELS_DIR, "crop_rf_model.pkl"))
joblib.dump(lc, os.path.join(MODELS_DIR, "label_encoder.pkl"))

print("Saved random_forest_model -> models/crop_rf_model.pkl")
print("Saved label encoder      -> models/label_encoder.pkl")
```
